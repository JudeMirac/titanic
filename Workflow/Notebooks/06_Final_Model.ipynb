```python
import json 
import joblib
import pandas as pd
from pathlib import Path 
from sklearn.linear_model import LogisticRegression
```

```python
# load data 

# current working directory 
dir = Path.cwd()

# find the data 
data = dir.parent / 'Data' / 'linear.csv'
target = dir.parent / 'Data' / 'target.csv'

# store into dataframe 
data = pd.read_csv(data)
target = pd.read_csv(target)

# review 
print(f'Dataset Preview: {data.head()}')
print(f'Target Class Preview: {target.head()}')
```

```python
# independent varibales 
X = data.copy()

# dependent variable 
Y = target['Survived']
```

```python
# train final model  
from sklearn.linear_model import LogisticRegression 

# instantiation process
final_model = LogisticRegression(C=1, penalty= 'l1', class_weight= 'balanced', solver = 'liblinear', random_state= 42)

# fit the data onto model 
final_model.fit(X, Y)
```

```python
# save model 
model_path =  dir.parent / 'Final_Model.pkl'

joblib.dump(final_model, model_path)
```

```python
# model metadata 

metadata = { 
    'model_name': 'Logistic Regression', 
    'model_type': 'Binary Classification', 
    'target': 'Survived', 
    'positive_class': 1, 
    'features': list(X.columns), 
    'selected_threshold': 0.45, 
    'hyperparameters': {
        'C': 1, 
        'Solver': 'liblinear', 
        'Class_Weight': 'Balanced', 
        'random_state': 42,
    }, 
    
    'evaluation_summary': {
        'test_accuracy': 0.8246, 
        'test_precision': 0.7807, 
        'test_racll': 0.8018, 
        'test_f1_score': 0.7911, 
        'selected_threshold': 0.45
    }
}

metadata_path = dir.parent / 'final_model_metadata.json' 

with open(metadata_path, 'w') as f:
    json.dump(metadata, f, indent = 4)

print(f'Final Model save to {model_path}')
print(f'Model metadata saved to {metadata_path}')
```

# Final Model Build

This phase trains the selected final model on the full curated linear dataset. Based on the model evaluation workflow, Logistic Regression was selected as the final model because it provided competitive performance, stronger interpretability, stable calibration behavior, and clearer error patterns than SVM.

The model was trained using the final selected hyperparameters and saved. The selected classification threshold of 0.45 was also stored in the model metadata because it produced the strongest balance between precision and recall during threshold tuning.

The saved model can be used for future predictions, while the metadata file documents the selected features, hyperparameters, threshold, and evaluation summary.
