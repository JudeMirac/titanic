# Feature Testing and Predictive Power Review

This notebook evaluates the predictive strength of engineered features created during the EDA phase. The goal is to determine whether interaction and ratio-based features add useful signal for predicting Titanic survival or introduce unnecessary noise.

Predictive Power Score (PPS) is used to compare engineered features against their original component variables. If an engineered feature has weaker predictive power than the variables used to create it, the feature may be adding noise rather than improving the dataset.

The results from this notebook help guide feature selection for downstream hypothesis testing, modeling, and model evaluation.

```python
# Load libraries
import pandas as pd # data manipulation
from pathlib import Path # located directory
```

```python
# upload dataset with engineered features

# locate the directory
dir = Path.cwd()

# point to the data file 
data_path = dir.parent / 'Data' / 'uncut_feat.csv'

# load data 
df1 = pd.read_csv(data_path)
```

```python
# before traing let's view which features are important in predicitng the target variable 

# Download ppscore 
!pip install ppscore

# import ppscore
import ppscore as pps

# convert target variable to object type 
df1['Survived'] = df1['Survived'].astype(str)
```

# Predictive Power Score Setup

Predictive Power Score was used to measure how strongly each feature predicts the survival target. Unlike correlation, PPS can detect both linear and non-linear predictive relationships, making it useful for evaluating engineered interaction features.

The target variable was converted to a categorical format so PPS could evaluate survival as a classification outcome.

## Interaction Feature Testing

This section compares each engineered interaction feature against the original variables used to create it. The purpose is to determine whether combining variables improves predictive signal or weakens it.

An interaction feature is considered useful when its PPS score is stronger than, or meaningfully competitive with, the individual input variables. If the interaction score is weaker than the original variables, the feature may be creating informational dilution or unnecessary noise.

```python
# calculate PPS for the engineered feature
interaction1 = pps.score(df1, 'interaction1', 'Survived')
# Calculate PPS for individual variables 
score_sibsp = pps.score(df1, 'SibSp', 'Survived') 
score_parch = pps.score(df1, 'Parch', 'Survived')
# print results 
score_sibsp, score_parch, interaction1

# if the score is higher than the interaction feature, then it must be removed due to iniformational dilution
# (In otherwords, combining the variables togther may have weakend the predicitive power of the other leading to nosie creation)
```

```python
# calculate PPS for engineered features
interaction2 = pps.score(df1, 'interaction2', 'Survived')

# calculate PPS for individually features (Note: no need to implement variables already calculated upstream!
score_fare = pps.score(df1, 'Fare', 'Survived') 

# Print results 
score_fare, score_sibsp, interaction2
```

# Feature Testing Summary

The feature testing phase showed that not every engineered feature improved predictive signal. Some interaction and ratio features performed competitively, while others appeared weaker than their original component variables.

This review helped identify which engineered features should be carried forward and which features may require removal during hypothesis testing or model evaluation. The goal is to reduce noise, avoid redundant predictors, and create cleaner datasets for both linear and non-linear models.

These findings were used to support the later creation of separate linear and non-linear datasets for downstream modeling.
